### Evidence Standards

Record both collection date and publication or effective date
where available.

Preserve original observations and maintain a source reference.

Separate verified facts, company claims, customer opinions,
assumptions, and recommendations.

Analyze company selected testimonials separately from independent reviews.

Collect only through permitted methods.

Avoid collecting reviewer names and unnecessary personal information.

### Review Sampling

Document the selection method, time window, available review count,
and collected review count for each source.

Do not select only positive reviews or severe complaints.

Keep brand level Reddit discussions separate from branch level reviews
unless a branch is explicitly identified.

Treat inaccessible reviews as missing coverage rather than evidence
that customers have no complaints.

## 6. Proposed Metrics

| Metric | Definition | Interpretation |
| :--- | :--- | :--- |
| Comparable basket cost | Total regular price of a defined set of equivalent products | Cost of a similar purchase occasion |
| Price per 100 ml | Price divided by verified volume in ml, multiplied by 100 | Volume adjusted price, not a complete quality measure |
| Promotional saving percentage | Regular qualifying basket cost less promotional cost, divided by regular cost, multiplied by 100 | Saving after meeting offer conditions |
| Loyalty reward value | Comparable reward value divided by qualifying spend, multiplied by 100 | Reward value relative to required spending |
| Review topic share | Reviews mentioning a topic divided by eligible sampled reviews, multiplied by 100 | Topic frequency within the sample |
| Negative review share | Reviews classified as negative divided by reviews with valid sentiment labels, multiplied by 100 | Negative sentiment within the sample |
| Rating distribution | Count and share of each rating within a platform | Distribution of submitted ratings |
| Verified price change | Current comparable price less historical price, divided by historical price, multiplied by 100 | Nominal price movement |
| Observed service time | Minutes from a defined starting event to a defined ending event | Service experience during sampled visits |

### Comparison Rules

Compare equivalent products, sizes, channels, and store formats.

Keep regular prices separate from promotional prices.

Do not assume cup capacity equals the actual beverage volume.

Do not combine ratings from different platforms into one average.

Distinguish cumulative platform ratings from ratings in the selected
review period.

A review may mention several topics, so topic shares may total
more than 100 percent.

Record the denominator and missing observations for each metric.

Do not interpret branch counts as market share.

Do not infer competitor sales or profitability from review counts,
followers, or store appearance.

## 7. Analysis Plan

### Descriptive Exploration

Examine counts, missing values, duplicates, central tendency,
variation, distributions, and outliers.

Explore pricing, menu coverage, promotions, store features,
ratings, review topics, and geographic coverage.

### Relationships

Examine relevant relationships between variables.

Examples include price and rating, store format and service feedback,
and channel and product availability.

Association does not establish causation.

### Hypothesis Testing

Statistical tests will be selected after reviewing the data,
sampling design, independence, and assumptions.

| Candidate question | Data requirement |
| :--- | :--- |
| Do comparable drink prices differ between selected competitor groups? | Matched product definitions and a defensible sampling approach |
| Does negative service feedback differ by store format? | Consistent review coding and treatment of reviews clustered within branches |
| Have comparable product prices changed since 2024? | Reliable matched historical and current observations |

These questions are provisional and do not predetermine the test.

### Required Reporting for Each Valid Test

1. Business question and linked research question ID
2. Null and alternative hypotheses
3. Sampling limitations and assumption checks
4. Alpha level, provisionally 0.05
5. Appropriate test and reference distribution
6. Test statistic and p value
7. Effect size and uncertainty where appropriate
8. Critical value visualization with rejection regions
9. LCV and UCV where applicable
10. Reject the null or fail to reject the null
11. Business interpretation and limitations

Use only the applicable boundary for a one sided test.
Do not invent a second critical value.

Investigate outliers before deciding whether to retain, correct,
or exclude them. Document all exclusions.

### Predictive Modeling

Modeling is conditional on sufficient data and a useful prediction target.

Any model must be compared with a simple baseline and evaluated
using a suitable validation approach.

Public competitor observations alone may not support a reliable
prediction of sales or profitability.
