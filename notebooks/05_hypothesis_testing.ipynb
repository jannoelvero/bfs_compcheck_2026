```python
from pathlib import Path

import pandas as pd
from IPython.display import display

# Locate the EDA outputs.

project_root = Path.cwd().resolve()

while (
    not (
        project_root / "reports" / "eda" / "hypothesis_readiness.csv"
    ).is_file()
    and project_root != project_root.parent
):
    project_root = project_root.parent

readiness_path = (
    project_root / "reports" / "eda" / "hypothesis_readiness.csv"
)

if not readiness_path.is_file():
    raise FileNotFoundError(
        "Run Notebook 04 and save its EDA tables first."
    )

# Load and check the saved assessment.

hypothesis_readiness = pd.read_csv(
    readiness_path,
    dtype="string",
)

required_columns = [
    "research_question_id",
    "proposed_relationship",
    "available_evidence",
    "current_analysis",
    "test_status",
    "remaining_requirement",
]

missing_columns = [
    column
    for column in required_columns
    if column not in hypothesis_readiness.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

if hypothesis_readiness["research_question_id"].isna().any():
    raise ValueError("Missing research question identifiers.")

if hypothesis_readiness["research_question_id"].duplicated().any():
    raise ValueError("Repeated research question identifiers.")

with pd.option_context("display.max_colwidth", None):
    display(hypothesis_readiness)

print("Research questions loaded:", len(hypothesis_readiness))
print("No statistical tests performed. No files changed.")
```
