## Notebook 07: Predictive Modeling Readiness

### Purpose

Determine whether the collected evidence supports a useful
prediction linked to the business objectives.

### Requirements

A predictive model needs a clearly defined outcome, suitable
observations and a defensible validation approach.

Inputs must be available at the time a prediction would be made.

Evaluate performance against a simple baseline.
Keep related observations together when separating training
and evaluation data.

### Current Limitations

The project contains selected public competitor observations.

It does not contain sales outcomes, costs, conversion outcomes
or a repeated series of comparable prices.

Customer sentiment labels remain provisional and come from
36 platform selected comments.

Hot and iced menu records repeat price patterns. Randomly
splitting these records could overstate model performance.

### Decision Rule

Train a model only when its prediction target, available inputs
and evaluation design support a useful business decision.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

# Locate the saved feature tables.

project_root = Path.cwd().resolve()

while (
    not (
        project_root
        / "data"
        / "processed"
        / "features"
        / "feature_dictionary.csv"
    ).is_file()
    and project_root != project_root.parent
):
    project_root = project_root.parent

features_path = project_root / "data" / "processed" / "features"

if not (features_path / "feature_dictionary.csv").is_file():
    raise FileNotFoundError(
        "Run Notebook 06 and save its feature outputs first."
    )

# Assess possible prediction targets.

modeling_readiness = pd.DataFrame(
    [
        {
            "prediction_target": "Future comparable menu price",
            "research_question_ids": "RQ01;RQ10",
            "evidence_limitation": (
                "No repeated series of verified comparable prices."
            ),
            "decision": "Deferred",
            "required_data": (
                "Comparable dated prices across multiple periods, "
                "with enough history for chronological evaluation."
            ),
        },
        {
            "prediction_target": "Customer comment sentiment",
            "research_question_ids": "RQ05",
            "evidence_limitation": (
                "Only 36 platform selected comments. Labels remain "
                "provisional and related comments may share a branch."
            ),
            "decision": "Deferred",
            "required_data": (
                "A larger documented sample, reviewed labels and "
                "evaluation appropriate to the intended prediction use."
            ),
        },
        {
            "prediction_target": "Sales or franchise conversion",
            "research_question_ids": "RQ08;RQ09",
            "evidence_limitation": (
                "No recorded sales outcomes or lead conversion outcomes."
            ),
            "decision": "Deferred",
            "required_data": (
                "Authorized dated transactions or lead histories, "
                "outcomes and inputs available before each outcome."
            ),
        },
        {
            "prediction_target": "Price difference from its component prices",
            "research_question_ids": "RQ01",
            "evidence_limitation": (
                "The outcome is exactly calculated from the proposed inputs."
            ),
            "decision": "Use direct calculation",
            "required_data": (
                "No model required. Retain the documented price formula."
            ),
        },
    ]
)

with pd.option_context("display.max_colwidth", None):
    display(modeling_readiness)

print("Prediction targets assessed:", len(modeling_readiness))
print("No model trained. No performance scores calculated.")
print("No files changed.")
```

## Modeling Readiness Conclusion

Three proposed prediction targets remain deferred.
The fourth is an arithmetic calculation that requires no model.

No model was trained, and no predictive performance is claimed.

Future price prediction requires comparable observations across
multiple periods and chronological evaluation.

Sentiment prediction requires reviewed labels and a larger
documented sample suited to the intended use.

Sales or franchise conversion prediction would extend the
current research scope. It requires a documented amendment
and authorized outcome data before development.

The current study will proceed to descriptive visualization
and qualified business recommendations.

```python
from datetime import datetime
import shutil

modeling_path = project_root / "reports" / "modeling"
modeling_path.mkdir(parents=True, exist_ok=True)

output_path = modeling_path / "modeling_readiness.csv"

if output_path.is_file():
    backup_label = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
    backup_path = modeling_path / "backups" / backup_label
    backup_path.mkdir(parents=True, exist_ok=True)

    shutil.copy2(output_path, backup_path / output_path.name)

modeling_readiness.to_csv(output_path, index=False)

print("Assessments saved:", len(modeling_readiness))
print("Saved:", output_path)
print("No model trained. Raw and prepared datasets unchanged.")
```
