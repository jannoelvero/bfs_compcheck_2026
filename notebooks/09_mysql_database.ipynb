```python
from pathlib import Path

import pandas as pd
from IPython.display import display

project_root = Path.cwd().resolve()

while (
    not (
        project_root
        / "data/processed/quality_reports/field_types.csv"
    ).is_file()
    and project_root != project_root.parent
):
    project_root = project_root.parent

schema_path = (
    project_root
    / "data/processed/quality_reports/field_types.csv"
)

if not schema_path.is_file():
    raise FileNotFoundError("Run this notebook from inside the project.")

field_types = pd.read_csv(schema_path, dtype="string")

display(field_types)
```

```python
for dataset_name in field_types["dataset"].drop_duplicates():
    print("\nTable:", dataset_name)

    display(
        field_types.loc[
            field_types["dataset"].eq(dataset_name),
            ["field", "pandas_dtype"]
        ].reset_index(drop=True)
    )
```
